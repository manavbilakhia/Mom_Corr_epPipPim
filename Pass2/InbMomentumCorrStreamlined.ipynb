```python
def fit2d(h2, sec):
    import numpy

    xlims = numpy.arange(0.5, 5.5, 0.5)
    gx = h2.GetXaxis()
    hys = []
    gr = ROOT.TGraphErrors()
    gr.SetMarkerStyle(20)

    for x1, x2 in zip(xlims[:-1], xlims[1:]):
        ib1, ib2 = gx.FindBin(x1), gx.FindBin(x2)
        hy = h2.ProjectionY(f"hy_{ib1}_{ib2}", ib1, ib2)
        hy.SetDirectory(0)
        hys.append(hy)

        # If there's very little data, skip fitting
        if hy.GetEntries() < 5:
            continue

        # Estimate peak position and amplitude
        mx = hy.GetBinCenter(hy.GetMaximumBin())
        max_val = hy.GetMaximum()
        if max_val <= 0:
            continue

        # Define total function: gaus(0) + (pol1 + constant offset) background
        fy = ROOT.TF1("fy", "gaus(0) + pol1(3) + abs([5])", mx - 0.8, mx + 0.8)

        # Initial parameter guesses
        fy.SetParameter(0, max_val)    # Gaussian amplitude
        fy.SetParameter(1, mx)         # Gaussian mean
        fy.SetParameter(2, 0.1)        # Gaussian sigma
        fy.SetParameter(3, max_val * 0.2)  # Linear background offset
        fy.SetParameter(4, 0.0)        # Linear slope
        fy.SetParameter(5, max_val * 0.1)  # Constant offset to ensure positivity

        # Parameter limits to ensure positivity
        fy.SetParLimits(0, 0, 1e9)  # Gaussian amplitude >= 0
        fy.SetParLimits(1, mx - 0.5, mx + 0.5)
        fy.SetParLimits(2, 0.001, 2.0)  # Gaussian sigma > 0
        fy.SetParLimits(3, 1e-3, 1e9)   # Linear offset > 0 (strictly positive)
        fy.SetParLimits(4, -1e3, 1e3)   # Linear slope constrained
        fy.SetParLimits(5, 1e-3, 1e9)   # Additional constant offset > 0 to force positivity

        # Attempt the fit with retries if needed
        for _ in range(3):  # Retry up to 3 times
            fit_result = hy.Fit(fy, "QRS")
            status = fit_result.Status()
            if status == 0:
                break  # Successful fit
            # Increase polynomial terms if fit fails
            fy.SetParameter(3, fy.GetParameter(3) * 1.2)
            fy.SetParameter(5, fy.GetParameter(5) * 1.2)

        if status != 0:
            print(f"Warning: Fit failed for slice {x1}-{x2}, skipping.")
            continue

        # If successful, refine the final fit range around ±3 sigma
        mu = fy.GetParameter(1)
        sig = abs(fy.GetParameter(2))
        if sig < 0.001:
            sig = 0.01
        fy.SetRange(mu - 3*sig, mu + 3*sig)
        hy.Fit(fy, "QRS")

        # Keep track in TGraph
        hy.SetTitle(f"{x1:.2f}<p<{x2:.2f} sec{sec:.0f}")
        gr.SetPoint(gr.GetN(), 0.5*(x1 + x2), fy.GetParameter(1))
        gr.SetPointError(gr.GetN() - 1, 0, fy.GetParError(1))

        # Separate signal + background
        f_gaus = ROOT.TF1("f_gaus", "gaus(0)", mu - 3*sig, mu + 3*sig)
        f_gaus.SetParameters(fy.GetParameter(0), fy.GetParameter(1), fy.GetParameter(2))
        f_gaus.SetLineColor(ROOT.kGreen)

        # pol1 background with constant offset
        f_bkg = ROOT.TF1("f_bkg", "pol1(0) + abs([2])", mu - 3*sig, mu + 3*sig)
        f_bkg.SetParameters(
            fy.GetParameter(3),
            fy.GetParameter(4),
            fy.GetParameter(5)
        )
        f_bkg.SetLineColor(ROOT.kBlack)

        # Total in red
        fy.SetLineColor(ROOT.kRed)

        hy.GetListOfFunctions().Clear()
        hy.GetListOfFunctions().Add(fy)
        hy.GetListOfFunctions().Add(f_gaus)
        hy.GetListOfFunctions().Add(f_bkg)

    # Attach histograms and TGraph to h2
    setattr(h2, "hys", hys)
    setattr(h2, "gr", gr)

    # Optionally fit the extracted means vs momentum
    fdp = ROOT.TF1("fdp", "pol2(0)", 0.5, 5)
    fdp.SetLineColor(ROOT.kMagenta)
    gr.Fit(fdp, "QR")
    setattr(h2, "fdp", fdp)
```
